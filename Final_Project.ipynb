```python
# =========================================
# AIR QUALITY ANALYSIS & VISUALIZATION
# =========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Clean look, no background grid
sns.set(style="white")

# -----------------------------------------
# 1. Load Dataset
# -----------------------------------------

df = pd.read_csv("AirQualityUCI.csv", sep=',')

print("Dataset loaded successfully ✅")
print(df.head())

# -----------------------------------------
# 2. Data Cleaning
# -----------------------------------------
# -200 ka means missing value
df.replace(-200, np.nan, inplace=True)


df.dropna(inplace=True)

print("\nAfter cleaning, dataset shape:", df.shape)

# -----------------------------------------
# 3. Date & Time Conversion
# -----------------------------------------
# Date aur Time ko proper format me convert karna
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S').dt.time

# -----------------------------------------
# 4. CO Pollution Trend Over Time
# -----------------------------------------
# Ye graph batata hai ki CO pollution time ke sath kaise change hota hai
plt.figure(figsize=(12,5))
plt.plot(df['Date'], df['CO(GT)'], color='green')
plt.title("CO Pollution Level Over Time")
plt.xlabel("Date")
plt.ylabel("CO Level (mg/m³)")
plt.grid(False)
plt.show()

# -----------------------------------------
# 5. Temperature Distribution
# -----------------------------------------
# Ye graph dikhata hai temperature kis range me zyada raha
plt.figure(figsize=(8,5))
plt.hist(df['T'], bins=25, color='orange')
plt.title("Temperature Distribution")
plt.xlabel("Temperature (°C)")
plt.ylabel("Number of Records")
plt.grid(False)
plt.show()

# -----------------------------------------
# 6. Correlation Heatmap (Only Numeric Data)
# -----------------------------------------

numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(), cmap='coolwarm')
plt.title("Relationship Between Air Pollutants")
plt.show()

# -----------------------------------------
# 7. Scatter Plot: CO vs NO2
# -----------------------------------------

plt.figure(figsize=(8,5))
plt.scatter(df['CO(GT)'], df['NO2(GT)'], color='purple')
plt.title("Relation Between CO and NO2")
plt.xlabel("CO (mg/m³)")
plt.ylabel("NO2 (µg/m³)")
plt.grid(False)
plt.show()

# -----------------------------------------
# 8. Boxplot: Pollution Comparison
# -----------------------------------------

plt.figure(figsize=(10,5))
sns.boxplot(data=df[['CO(GT)', 'NO2(GT)', 'NOx(GT)']])
plt.title("Comparison of Different Pollutants")
plt.ylabel("Pollution Level")
plt.show()

# -----------------------------------------
# 9. Final Conclusion
# -----------------------------------------
print("\nCONCLUSION:")
print("• Air pollution levels change over time.")
print("• Some pollutants increase together, showing a strong relationship.")
print("• Graphs help in understanding air pollution data easily.")
print("\nPROJECT COMPLETED SUCCESSFULLY 🎉")
```
